```python
# Read the dataset and store it as data2026
data2026=pd.read_csv('2026PartYear.csv', encoding='utf-8') 
```

```python
# Check periods available in each dataset
print(data2026['Period'].unique())
print(data2025['Period'].unique())
```

```python
# keep the same periods in 2025 as the 2026 data

data2025_sameperiod=data2025[data2025['Period'].isin(data2026['Period'].unique())]
```

```python
print(data2025_sameperiod['Period'].unique())
```

```python
# Filter Total U.S. data

US2025 = data2025_sameperiod[data2025_sameperiod['Geography'] == 'Total U.S.']

US2026 = data2026[data2026['Geography'] == 'Total U.S.']
```

```python
# Total U.S. avocado volume

Volume2025 = US2025['Total Bulk and Bags'].sum()
Volume2026 = US2026['Total Bulk and Bags'].sum()

print("2025 Volume:", Volume2025)
print("2026 Volume:", Volume2026)
```

```python
# Percent change in Total U.S. volume

VolumeChange = ((Volume2026 - Volume2025) / Volume2025) * 100
print("Volume Change:", round(VolumeChange, 1), "%")
```

```python
# Calculate weighted average selling price

ASP2025 = ((US2025['ASP Current Year'] * US2025['Total Bulk and Bags']).sum()/US2025['Total Bulk and Bags'].sum())
ASP2026 = ((US2026['ASP Current Year'] * US2026['Total Bulk and Bags']).sum()/US2026['Total Bulk and Bags'].sum())

print("2025 ASP: $", round(ASP2025, 2))
print("2026 ASP: $", round(ASP2026, 2))
```

```python
# Calculate percent change in ASP

ASPChange = ((ASP2026 - ASP2025) / ASP2025) * 100

print("ASP Change:", round(ASPChange, 1), "%")
```

```python
# Filter conventional avocados
Conventional2025 = US2025[US2025['Type'] == 'Conventional']

Conventional2026 = US2026[US2026['Type'] == 'Conventional']
```

```python
# Conventional avocado volume

ConvVolume2025 = Conventional2025['Total Bulk and Bags'].sum()
ConvVolume2026 = Conventional2026['Total Bulk and Bags'].sum()

ConvVolumeChange = ((ConvVolume2026 - ConvVolume2025) / ConvVolume2025) * 100

print("Conventional Volume Change:", round(ConvVolumeChange, 1), "%")
```

```python
# Conventional weighted average selling price

ConvASP2025 = ((Conventional2025['ASP Current Year'] * Conventional2025['Total Bulk and Bags']).sum()/Conventional2025['Total Bulk and Bags'].sum())
ConvASP2026 = ((Conventional2026['ASP Current Year'] * Conventional2026['Total Bulk and Bags']).sum()/Conventional2026['Total Bulk and Bags'].sum())
ConvASPChange = ((ConvASP2026 - ConvASP2025) / ConvASP2025) * 100

print("Conventional ASP Change:", round(ConvASPChange, 1), "%")
```

```python
# Filter organic avocados

Organic2025 = US2025[US2025['Type'] == 'Organic']
Organic2026 = US2026[US2026['Type'] == 'Organic']
```

```python
# Organic avocado volume

OrganicVolume2025 = Organic2025['Total Bulk and Bags'].sum()
OrganicVolume2026 = Organic2026['Total Bulk and Bags'].sum()

OrganicVolumeChange = ((OrganicVolume2026 - OrganicVolume2025)/OrganicVolume2025) * 100

print("Organic Volume Change:", round(OrganicVolumeChange, 1), "%")
```

```python
# Organic weighted average selling price

OrganicASP2025 = ((Organic2025['ASP Current Year'] * Organic2025['Total Bulk and Bags']).sum()/Organic2025['Total Bulk and Bags'].sum())

OrganicASP2026 = ((Organic2026['ASP Current Year'] * Organic2026['Total Bulk and Bags']).sum()/Organic2026['Total Bulk and Bags'].sum())

OrganicASPChange = ((OrganicASP2026 - OrganicASP2025) / OrganicASP2025) * 100

print("Organic ASP Change:", round(OrganicASPChange, 1), "%")
```

```python
# Calculate total bagged avocado volume
Bagged2025 = US2025['TotalBagged'].sum()
Bagged2026 = US2026['TotalBagged'].sum()

print("2025 Bagged Volume:", Bagged2025)
print("2026 Bagged Volume:", Bagged2026)
```

```python
# Calculate percent change in bagged volume
BaggedChange = ((Bagged2026 - Bagged2025) / Bagged2025) * 100

print("Bagged Volume Change:",
      round(BaggedChange, 1), "%")
```

```python
# Calculate bagged avocados as a percent of total volume

BagShare2025 = (Bagged2025 / Volume2025) * 100
BagShare2026 = (Bagged2026 / Volume2026) * 100

print("2025 Bagged Share:", round(BagShare2025, 1), "%")

print("2026 Bagged Share:", round(BagShare2026, 1), "%")
```

```python
# Calculate conventional bagged avocado growth

ConvBagged2025 = Conventional2025['TotalBagged'].sum()
ConvBagged2026 = Conventional2026['TotalBagged'].sum()

ConvBaggedChange = ((ConvBagged2026 - ConvBagged2025) / ConvBagged2025) * 100

print("Conventional Bagged Volume Change:", round(ConvBaggedChange, 1), "%")
```

```python
# List of major U.S. regions
regions = ['California','Great Lakes','Midsouth','Northeast','Plains','South Central','Southeast','West']
```

```python
# Calculate volume change for each region
for region in regions:

    Region2025 = data2025_sameperiod[
        data2025_sameperiod['Geography'] == region
    ]

    Region2026 = data2026[
        data2026['Geography'] == region
    ]

    RegionVolume2025 = Region2025['Total Bulk and Bags'].sum()
    RegionVolume2026 = Region2026['Total Bulk and Bags'].sum()

    RegionChange = (
        (RegionVolume2026 - RegionVolume2025)
        / RegionVolume2025
    ) * 100

    print(region, round(RegionChange, 1), "%")
```

```python
# Calculate ASP change for each region
for region in regions:

    Region2025 = data2025_sameperiod[
        data2025_sameperiod['Geography'] == region
    ]

    Region2026 = data2026[
        data2026['Geography'] == region
    ]

    RegionASP2025 = (
        (Region2025['ASP Current Year'] *
         Region2025['Total Bulk and Bags']).sum()
        / Region2025['Total Bulk and Bags'].sum()
    )

    RegionASP2026 = (
        (Region2026['ASP Current Year'] *
         Region2026['Total Bulk and Bags']).sum()
        / Region2026['Total Bulk and Bags'].sum()
    )

    RegionASPChange = (
        (RegionASP2026 - RegionASP2025)
        / RegionASP2025
    ) * 100

    print(region, round(RegionASPChange, 1), "%")
```
